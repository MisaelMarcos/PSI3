```python
df_pernambuco = pd.read_csv('../dados/df_pernambuco.csv')
```

Filtro Apenas Recife

```python
df_recife = df_pernambuco[df_pernambuco['ESTACAO'] == 'RECIFE']
df_recife
```

Limpeza de colunas

```python
colunas_inuteis = ['DATA DE FUNDAC?O', 'RADIACAO GLOBAL (Kj/m²)', 'DATA DE FUNDACAO', 'ALTITUDE', 'LONGITUDE', 'LATITUDE', 'CODIGO (WMO)', 'ESTACAO', 'UF', 'REGIAO', 'RADIACAO GLOBAL (KJ/m²)']
df_recife.drop(columns = colunas_inuteis, inplace=True)
df_recife.reset_index(inplace=True, drop=True)
df_recife
```

```python
df_recife.info()
```

Ajuste do tipo da coluna Data

```python
df_recife["Data"] = pd.to_datetime(df_recife["Data"], format="%Y/%m/%d", errors="coerce")
```

```python
df_recife.to_csv('../dados/df_recife.csv')
```
